Lab 1: List Comprehension

```python
#Sqaures of number 0 to 4
squares = []
for x in range(5):
    squares.append(x**2)
print(squares)
```

```python
# Using list comprehension
sqaures = [x**2 for x in range(5)]
print(squares)
```

```python
#List of numbers from 1 to 10
numbers = [x for x in range(1,11)]
print(numbers)
```

```python
even = [x for x in numbers if x%2 == 0]
print(even)
```

```python
# Reversing each string in a list
words = ["hello", "world", "python"]
reversed_words = [word[::-1] for word in words]
print(reversed_words)
```

```python
# Extracting vowels from a given string
sentence = "Hello, how are you today?"
vowels = [char for char in sentence if char.lower() in 'aeiou']
print(vowels)
```

```python
# Create a list of sqaures of odd numbers from 1 to 10
odd = [x**2 for x in range(1,11) if x%2!=0]
print(odd)
```

Lab 2: Reduce Function

```python
from functools import reduce
# Sample list
numbers = [1, 2, 3, 4, 5]
# Using reduce to calculate the sum of the numbers in the list
sum_of_numbers = reduce(lambda x, y: x + y, numbers)
print(sum_of_numbers)
# Output: 15

#Without lambda
def add(x,y):
    return x+y
numbers = [1, 2, 3, 4, 5]
result = reduce(add, numbers)
```

```python
#Take the list as a input and the output should be the multiplication of all the elements
from functools import reduce
l=[1,2,3,6]
result = reduce(lambda a,b:a*b,l)
print("Product: ", result)
```

```python
#Find out the maximum value from the list without using max function
from functools import reduce
l=[3,8,2,12,34]
reduce(lambda a,b: a if a>b else b,l)
```

```python
#Find out the minimum value from the list without using min function
from functools import reduce
l=[7,23,45,11,4,19]
reduce(lambda a,b:a if a<b else b,l)
```

```python
#Calculate the factorial of a given number n by using reduce and lambda function
from functools import reduce
n=int(input('n='))
fact=reduce(lambda a,b:a*b,range(1,n+1))
print(fact)
```

```python
#Calculate the multiplication of all the even numbers with in a given list.
from functools import reduce
l=[12,24,11,6,8,16,17,13]
result = reduce(lambda a,b:a*b,[i for i in l if i%2==0])
print(result)
```
