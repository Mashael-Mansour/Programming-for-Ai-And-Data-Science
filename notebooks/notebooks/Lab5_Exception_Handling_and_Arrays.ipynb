```python
#Handle specific errors
try:
    a = int(input("a = "))
    b = int(input("b = "))
    c = a/b
    print("Result:", c)

except ZeroDivisionError:
    print("Make sure b is not 0.")

except ValueError:
    print("Enter integer values only! ")

except Exception:
    print("Error Ocurred: ")
```

```python
try:
    a=int(input('a='))
    b=int(input('b='))
    c=a/b

except ZeroDivisionError as z:
    pass

except ValueError as v:
    print(v)

except Exception as e:
    print(e)

else:
    print('The answer is: ', c)

finally:
    print('Program End!')
```

```python
try:
    numbers = [10,20,30,40,50]
    index = int(input("Enter list index (0-4): "))
    print("Value at index: ", numbers[index])

except ValueError as v:
    print("Enter an integer!", v)

except IndexError as i:
    print("Number range 0 to 4!", i)
```

```python
def is_even(lst):
    even = []
    for i in lst:
        if i % 2 == 0:
            even.append(i)
    return even
print(is_even([1,2,3,4,5,6]))
print(is_even(23))
```

```python
def is_even(lst):
    even = []
    try:
        for i in lst:
            if i % 2 == 0:
                even.append(i)
        return even
    except TypeError:
        print("Please give a list!")

print(is_even([1,2,3,4,5,6]))
print(is_even(23))
```

Lab 2:Array

```python
#Arrays are faster, support matrix operations.
import numpy as np

arr_1d = np.array([1, 2, 3, 4, 5])
print(arr_1d)

arr_2d = np.array([[1, 2, 3], [4, 5, 6]])
print(arr_2d)
```

```python
# Accessing elements of the array
print(arr_1d[0]) # Accessing the first element
print(arr_1d[2]) # Accessing the third element
```

```python
# Accessing elements of the array
print(arr_2d[0][0]) # first row, first column
print(arr_2d[1][2]) # second row, third column
```

```python
print(arr_2d.reshape(1,6))
```

```python
# Performing operations on arrays
print(arr_1d+2)
print(arr_1d*2)
arr_sum2d = np.sum(arr_2d)
print("Sum of elements of 2D:", arr_sum2d)
```

```python
row = np.sum(arr_2d, axis=1)
print("Row wise sum:", row)
col = np.sum(arr_2d, axis=0)
print("Column wise sum:", col)
```
